```python

import pandas as pd
import numpy as np
from google.colab import files


uploaded = files.upload()

files_list = [
    "kolesa1.csv",
    "kolesa2.csv",
    "kolesa3.csv"
]


df_list = []

for file in files_list:
    temp = pd.read_csv(file)
    df_list.append(temp)

df = pd.concat(df_list, ignore_index=True)

print("Размер до cleaning:", df.shape)
display(df.head())


print("Колонки:")
print(df.columns)

print("\nПропуски до cleaning:")
print(df.isna().sum())

print("\nДубликаты до cleaning:")
print(df.duplicated().sum())



df_clean = df.copy()

df_clean.columns = df_clean.columns.str.strip()

for col in df_clean.columns:
    df_clean[col] = df_clean[col].astype(str).str.strip()
    df_clean[col] = df_clean[col].str.replace(r"\s+", " ", regex=True)

df_clean = df_clean.replace(
    ["", "nan", "NaN", "None", "none", "null", "NULL"],
    np.nan
)

df_clean = df_clean.dropna()

df_clean = df_clean.drop_duplicates()



df_clean["price"] = (
    df_clean["price"]
    .astype(str)
    .str.replace(r"[^\d]", "", regex=True)
)

df_clean["price"] = pd.to_numeric(df_clean["price"], errors="coerce")

df_clean = df_clean.dropna(subset=["price"])

df_clean = df_clean[
    (df_clean["price"] >= 100_000) &
    (df_clean["price"] <= 500_000_000)
]

df_clean["price"] = df_clean["price"].astype(int)


df_clean["year"] = pd.to_numeric(df_clean["year"], errors="coerce")

df_clean = df_clean.dropna(subset=["year"])

df_clean = df_clean[
    (df_clean["year"] >= 1980) &
    (df_clean["year"] <= 2026)
]

df_clean["year"] = df_clean["year"].astype(int)


valid_body_types = [
    "седан",
    "кроссовер",
    "купе",
    "универсал",
    "лифтбек",
    "хэтчбек",
    "кабриолет",
    "родстер"
]

df_clean = df_clean[df_clean["body_type"].isin(valid_body_types)]


df_clean["volume"] = (
    df_clean["volume"]
    .astype(str)
    .str.replace(",", ".", regex=False)
)

df_clean["volume"] = pd.to_numeric(df_clean["volume"], errors="coerce")

df_clean = df_clean.dropna(subset=["volume"])

df_clean = df_clean[
    (df_clean["volume"] > 0) &
    (df_clean["volume"] <= 8)
]


valid_fuels = [
    "бензин",
    "дизель",
    "газ-бензин",
    "гибрид",
    "электричество"
]

df_clean = df_clean[df_clean["fuel"].isin(valid_fuels)]


valid_gearboxes = [
    "автомат",
    "механика",
    "робот",
    "вариатор"
]

df_clean = df_clean[df_clean["gearbox"].isin(valid_gearboxes)]


valid_types = [
    "Б/у",
    "новый"
]

df_clean = df_clean[df_clean["type"].isin(valid_types)]


df_clean = df_clean[df_clean["title"].notna()]
df_clean = df_clean[df_clean["title"].str.startswith("BMW")]


df_clean["brand"] = df_clean["title"].str.split().str[0]
df_clean["model"] = df_clean["title"].str.replace("BMW", "", regex=False).str.strip()

df_clean = df_clean[df_clean["model"] != ""]


df_clean = df_clean[
    [
        "title",
        "brand",
        "model",
        "price",
        "year",
        "body_type",
        "volume",
        "fuel",
        "gearbox",
        "type"
    ]
]

df_clean = df_clean.drop_duplicates()

df_clean = df_clean.reset_index(drop=True)


print("Размер после cleaning:", df_clean.shape)
display(df_clean.head(20))

print("\nПропуски после cleaning:")
print(df_clean.isna().sum())

print("\nДубликаты после cleaning:")
print(df_clean.duplicated().sum())

print("\nТипы данных:")
print(df_clean.dtypes)

print("\nУникальные body_type:")
print(df_clean["body_type"].unique())

print("\nУникальные fuel:")
print(df_clean["fuel"].unique())

print("\nУникальные gearbox:")
print(df_clean["gearbox"].unique())

print("\nУникальные type:")
print(df_clean["type"].unique())


df_clean.to_csv("kolesa_cleaned_strict.csv", index=False, encoding="utf-8-sig")

files.download("kolesa_cleaned_strict.csv")
```
